```python
import torch
import numpy as np
```

```python
x=torch.arange(1.,10)
x
```

```python
x.shape
```

```python
x_reshape=x.reshape(1,9)
x_reshape
```

* view method doesn't make copy of tensor instead give you another view of the same tensor

```python
z=x.view(1,9)
z,z.shape
```

* changing view changes x

```python
z[:,0] =5
z,x
```

* stacking tensor on top of eachother

```python
x_stacked=torch.stack([x,x,x,x],dim=0)
x_stacked
```

* lets remove all single dim from a tensor
` torch.squeeze() `

```python
print(x_reshape)
x_reshape.shape
```

```python
x_squeezed=x_reshape.squeeze()
print(x_squeezed)
x_squeezed.shape
```

```python
x_unsqueezed=x_squeezed.unsqueeze(dim=0)
print(x_unsqueezed.shape)
x_unsqueezed
```

* rearrange the order of axes values with `torch.permute(input,dims)`

```python
x_original=torch.rand(size=(224,224,3))
print(f"Previous shape: {x_original.shape}")
# rearrage the axis
x_permuted=x_original.permute(2,0,1)
print(f"New shape: {x_permuted.shape}")
```
